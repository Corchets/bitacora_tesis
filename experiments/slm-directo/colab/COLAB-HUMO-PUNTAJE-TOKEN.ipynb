```python
# Deja el repositorio en el commit fijado (por tag, no por rama) y corre la prueba de humo.
import os, runpy, subprocess, pathlib
TAG = 'colab/143-humo'
COMMIT = '9a2692ee4ee9cf526531e96a1f27bdb67552fcc9'
REPO = pathlib.Path('/content/bitacora_tesis')
if not REPO.exists():
    subprocess.run(['git', 'clone', 'https://github.com/Corchets/bitacora_tesis.git', str(REPO)], check=True)
subprocess.run(['git', '-C', str(REPO), 'fetch', '--force', 'origin', f'refs/tags/{TAG}:refs/tags/{TAG}'], check=True)
apunta = subprocess.check_output(['git', '-C', str(REPO), 'rev-parse', f'{TAG}^{{commit}}'], text=True).strip()
if apunta != COMMIT:
    raise RuntimeError(f'El tag {TAG} apunta a {apunta}, no a {COMMIT}; detener y revisar.')
subprocess.run(['git', '-C', str(REPO), 'checkout', '--detach', COMMIT], check=True)
os.environ['COLAB_COMMIT'] = COMMIT
runpy.run_path(str(REPO / 'experiments/slm-directo/colab/humo_puntaje_token.py'), run_name='__main__')
```
